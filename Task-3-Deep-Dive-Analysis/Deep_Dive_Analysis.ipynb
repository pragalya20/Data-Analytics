# INTRODUCTION
## Deep-Dive Analysis: Regional Sales Performance

### Objective
The objective of this project is to perform a deep-dive business analysis on sales data and build an interactive dashboard for business intelligence and decision-making.

This analysis focuses on:
- Regional sales performance
- Product category trends
- Customer payment behavior
- Monthly sales trends

# IMPORT LIBRARIES

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("cleaned_sales_data.csv")

# Display first rows
df.head()
```

# KPI DEFINITIONS
## Core KPIs

### 1. Total Revenue
Measures total business sales.

### 2. Average Revenue per Order
Measures average customer spending.

### 3. Total Orders
Measures transaction volume.

### 4. Top Performing Region
Identifies strongest geographical market.

### 5. Top Product Category
Identifies highest revenue-generating category.

# REVENUE BY REGION

```python
region_revenue = df.groupby('Region')['Total Revenue'].sum()

region_revenue.plot(kind='bar', color='skyblue')

plt.title("Revenue by Region")
plt.ylabel("Revenue")
plt.show()
```

### Insight
North America generates the highest revenue among all regions.

# MONTHLY SALES TREND


```python
monthly_sales = df.groupby('Month')['Total Revenue'].sum()

monthly_sales.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")
plt.ylabel("Revenue")
plt.show()
```

### Insight
Sales fluctuate across months, with stronger performance in earlier months.

# PRODUCT CATEGORY ANALYSIS


```python
category_revenue = df.groupby('Product Category')['Total Revenue'].sum()

category_revenue.plot(kind='bar', color='orange')

plt.title("Revenue by Product Category")
plt.ylabel("Revenue")
plt.show()
```

### Insight
Electronics generate the highest revenue among all product categories.

# PAYMENT METHOD ANALYSIS



```python
payment_method = df['Payment Method'].value_counts()

payment_method.plot(kind='pie', autopct='%1.1f%%')

plt.title("Payment Method Analysis")
plt.ylabel("")
plt.show()
```

### Insight
Credit cards are the most preferred payment method among customers.

# Final Insights

- North America is the strongest revenue-generating region.
- Electronics dominate overall sales performance.
- Credit cards are the most preferred payment method.
- Monthly sales trends vary across periods.
- Interactive dashboards improve business decision-making.
